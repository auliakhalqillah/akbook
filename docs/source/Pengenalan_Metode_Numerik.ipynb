### Fungsi Matematika

Fungsi matematika adalah kumpulan variabel-variabel, koefisien, dan/atau konstanta yang memiliki implikasi hubungan dengan suatu variabel hasil. Fungsi matematika umumnya ditulis dengan simbol $f(x)$ atau $g(x)$. Pada penerapannya, fungsi matematika digunakan untuk menyelesaikan fenomena alam yang bersifat linear, seperti fenomena gerak translasi yang dipelajari di Fisika, dimana salah satu fungsi matematikanya adalah $s(t) = v . t$, dimana $s(t)$ adalah variabel jarak sebagai fungsi waktu. Persamaan ini menjelaskan perubahan jarak seiring bertambahnya waktu dengan kecepatan, $v$, yang konstan. Selain itu, fungsi matematika juga digunakan untuk membahasakan fenomena alam yang bersifat tidak linear (non-linear).

Faktanya, banyak sekali fungsi matematika yang menjelaskan fenomena alam secara sederhana. Contohnya adalah persamaan berikut

$x = x_0 + v_0t + \frac {1}{2} g t^2$

yang digunakan untuk menghitung jarak berdasarkan informasi kecepatan awal $v_0$ dan posisi awal $x_0$ dengan konstanta gravitasi $g$ selama waktu tertentu $t$.

Jika terdapat permasalahan, berapa lama waktu ($t$) yang diperlukan untuk mencapai jarak $x$ dengan posisi awal $x_0$, kecepatan awal $v_0$, dan percepatan gravitasi $g$?

Permasalahan ini bisa diselesaikan dengan pendekatan analitik menggunakan persamaan kuadrat

$x = \frac {-b \pm \sqrt{b^2 - 4ac}}{2a}$

karena persamaan pada contoh di atas memenuhi persamaan kuadrat yang memiliki nilai akar tertentu. Namun, persamaan tersebut juga bisa diselesaikan dengan menggunakan pendekatan numerik, melalui metode penentuan akar persamaan.

Pada konteks yang lebih luas, banyak fungsi matematika yang tidak bisa diselesaikan secara analitik (tidak memiliki rumus baku) dan dibutuhkan solusi atau pendekatan metode numerik, seperti persaman non-linear order tinggi (pangkat lebih dari 2), persamaan integral, dan diferensial yang kompleks.

Pada bab metode numerik ini akan dijelaskan metode-metode pendekatan matematika untuk menyelesaikan persoalan matematika yang tidak bisa diselesaikan menggunakan solusi analitik. Ada tiga topik yang akan dibahas:

1. Akar persamaan
2. Sistem persamaan linear
3. Interpolasi
4. Integrasi numerik
5. Diferensiasi numerik

Setiap topik tersebut dijelaskan dengan menggunakan contoh kasus sederhana dalam bidang ilmu fisika.

### Definisi Galat (Error)

Ketika mengimplementasikan metode numerik untuk menyelesaikan permasalahan matematika, akan selalu dikaitkan dengan seberapa besar galat (error) yang dihasilkan dari hasil numerik tersebut. Untuk menghitung seberapa besar galat  dari suatu proses numerik, maka dihitung dengan persamaan sebagai berikut

$error = |\frac {x_r - x_{old}}{x_r}|$

dimana

$x_{old} = x_i$ atau $x_{old} = x_f$ adalah nilai pendekatan sebelumnya, dan

$x_r$ adalah nilai pendekatan saat ini. Perhitungan galat ini bisa disesuaikan dengan kasus metode numerik yang menerapkan proses pengulangan (iterasi).

### Akar Persamaan

Akar persamaan adalah suatu nilai variabel $x$ yang menghasilkan suatu fungsi matematika $f(x) = 0$. Contoh sederhananya adalah persamaan kuadrat dengan bentuk $ax^2 + bx + c = 0$, dimana $a$, $b$, dan $c$ adalah koefisien.

Persamaan kuadrat ini memiliki sifat yang ditentukan dari diskriminan, $D = b^2 - 4ac$, dimana jika

- Jika D > 0: Memiliki dua akar real yang berbeda.
- Jika D = 0: Memiliki dua akar real yang sama (kembar).
- Jika D < 0: Memiliki dua akar imajiner (tidak ada akar real).

Bentuk persamaan kuadrat ini bisa diselesaikan secara analitik maupun numerik. Proses analitik dapat dihitung menggunakan persamaan

$x = \frac {-b \pm \sqrt{b^2 - 4ac}}{2a}$

namun juga bisa diselesaikan dengan pendekatan metode numerik.

Sebagai contoh, diketahui persamaan kuadrat

$x^2 + 2x - 3 = 0$, dimana nilai $a = 1$, $b = 2$, dan $c = -3$. Maka nilai akar persamaan melalui proses analitik adalah

```python
a, b, c = 1, 2, -3

# Diskriminan
D = b**2 - (4*a*c)
print("Diskriminan:", D)

x = (-1 * b + (b**2 - (4*a*c))**0.5) / (2 * a)
f = lambda x : x**2 + (2*x) - 3
print("Akar:", x)
print("f(Akar):", f(x))
```

Karena diskriminan ($D > 0$), maka persamaan kuadrat di atas memiliki dua nilai akar dengan nilai yang berbeda, yaitu $x_1 = -3$ dan $x_2 = 1$.

Solusi numerik untuk persamaan kuadrat (pangkat dua atau lebih) dapat dilakukan dengan menggunakan dua pendekatan, pendekatan tertutup dan pendekatan terbuka.
