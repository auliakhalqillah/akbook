## Diferensiasi Numerik

Diferensiasi merupakan suatu proses mengestimasi perubahan yang terjadi pada suatu fungsi terhadap suatu variabel, dimana variabel tersebut yang menjadi faktor pengikat fungsi tersebut. Contoh dalam fisika, dari suatu fungsi jarak, $s(t)$, ingin mengestimasi seberapa cepat suatu objek berpindah dari posisi awal ($s_i$) ke posisi akhir ($s_f$) pada selang waktu tertentu ($\Delta t$).

Metode diferensiasi numerik umumnya memilik tiga pendekatan, yaitu _forward difference_, _backward difference_, dan _central difference_. Pada artikel ini, hanya akan dibahas pendekatan _central difference_, karena metode ini merupakan gabungan antara kedua metode lainnya.

### Metode Central Difference

Secara konseptual, metode _central difference_ merupakan rata-rata selisih antara dua nilai fungsi yang berurutan, dimana dapat ditulis sebagai berikut

$f'(x_i) = \frac {f(x_{i+1}) - f(x_{i-1})}{2h}$

dimana $i = 1,2,3,...,n$, dan

$f(x_{i+1})$ adalah fungsi ketika $x_{i+1}$

$f(x_{i-1})$ adalah fungsi ketika $x_{i-1}$

$h$ adalah interval data yang bisa dihitung berdasarkan batas awal ($x_i$) dan batas akhir ($x_f$) serta banyak data, $n$, yang dimiliki sebagai berikut

$h = \frac {x_f - x_i} {n}$

Sebagai contoh, jika suatu objek bergerak dimana fungsi jaraknya adalah $s(t) = \frac {1}{2} a t^2$, dengan percepatan konstan $a=40$ $m/s^2$ selama $t = 50$ detik, seberapa cepat ($v$) objek tersebut mengalami perpindahan?

Berdasarkan contoh di atas, diketahui bahwa fungsinya adalah

$f(t) = s(t) = \frac {1}{2} a t^2$

Fungsi ini bisa diselesaikan secara analitik dengan menurunkan fungsi tersebut terhadap waktu, $t$, sehingga diperoleh fungsi turunannya sebagai berikut

$f'(t) = a t$

dimana fungsi turunan tersebut adalah kecepatan. Jika kita subtitusikan nilai percepatan dan waktu tempuh objek ke dalam fungsi tersebut, maka diperoleh kecepatannya sebagai berikut

$f'(t) = v(t) = 40 * 50 = 2000$ $m/s$

Sekarang coba kita estimasi besar kecepatan objek tersebut menggunakan pendekatan diferensiasi numerik (pendekatan _central difference_) sebagai berikut

Asumsi banyak data yang digunakan adalah $n = 10$, maka interval data (selang waktu) diperoleh

$dt = h = \frac {t_f - t_i} {n} = \frac {50-0}{10} = 5$ detik

dimana waktu awalnya, $t_i = 0$ detik yang kemudian akan berubah setiap penambahan selang waktu melalui

$t = t + dt$

untuk $t=5$ detik sebagai inisialisasi awal. Nilai inisial $t$ ini kemudian bisa disesuaikan.

Kita coba implementasikan metode numerik diferensiasi dalam program Python

```python
# Metode Diferensiasi, Central Difference Method

# f = lambda s,t: s/t
def f(a,t):
  if t == 0:
    return 0
  else:
    return 0.5 * a * t**2

a = 40 # m/s^2
n = 10

t = 5 # detik
ti = 0 # detik
tf = 50 # detik

# interval
dt = (tf-ti)/n

for i in range(n):
  fi = (f(a,t+dt) - f(a,t-dt))/(2*dt)
  t = t + dt

print(f"Interval: {dt} detik")
print(f"Pada waktu {t-dt} detik kecepatannya adalah {fi} m/s")
```

Selain metode _central difference_, metode _forward difference_ dan metode _backward difference_ juga merupakan metode yang digunakan pada diferensiasi numerik. Namun, metode _central difference_ sudah cukup mampu mewakili kedua metode lainnya, sehingga prosesnya lebih praktis.

Kedua metode tersebut diselesaikan dengan persamaab berikut

**Metode Forward Difference**

$f'(x_i) = \frac {f(x_{i+1}) - f(x_i)}{h}$

**Metode Backward Difference**

$f'(x_i) = \frac {f(x_i) - f(x_{i-1})}{h}$
